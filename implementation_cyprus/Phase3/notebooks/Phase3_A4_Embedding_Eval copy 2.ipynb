## Step 1: Load Data

```python
# ── Load embeddings — dimension-safe (handles mixed dims after multi-stage extraction) ──
#
# Strategy:
#   1. Load each fold independently, recording its dimension
#   2. If all folds share the same dim → average as before (original behavior)
#   3. If dims differ (e.g. Fold 0=768-dim, Fold 1=384-dim) →
#      project each fold to a common dim via PCA before averaging

from sklearn.decomposition import PCA

# ── Load per fold ──
fold_data  = {}   # fold → {key: np.array(dim)}
fold_dims  = {}   # fold → dim
fold_metas = {}   # fold → dict with stage info (if available)

print('=== Loading BSF embeddings per fold ===')
for fold, path in EMB_FILES.items():
    if not path.exists():
        print(f'  ⚠️  {fold}: NOT FOUND at {path}')
        continue
    d   = np.load(path, allow_pickle=True)
    dim = d[list(d.keys())[0]].shape[0]
    fold_data[fold]  = {k: d[k] for k in d.keys()}
    fold_dims[fold]  = dim

    meta_path = path.parent / (path.stem + '_meta.json')
    if meta_path.exists():
        fold_metas[fold] = json.load(open(meta_path))
        stage = fold_metas[fold].get('best_stage', 4)
        # Safely format diversity percentage
        div_val = fold_metas[fold].get('best_diversity_pct')
        if isinstance(div_val, (int, float)):
            div_display = f"{div_val:.1f}%"
        else:
            div_display = "?"
        print(f'✅ {fold}: {len(d.keys())} scans × {dim}-dim'
              f'(enc[{stage}], div={div_display})')
    else:
        print(f'  ✅ {fold}: {len(d.keys())} scans × {dim}-dim')

if not fold_data:
    raise RuntimeError('No embedding files found! Run Phase3_A5_Reextract_Embeddings first.')

# ── Dimension alignment ──
unique_dims = list(set(fold_dims.values()))
print(f'\nDimensions per fold: {fold_dims}')

if len(unique_dims) == 1:
    # All same dim → average directly (original behavior)
    COMMON_DIM = unique_dims[0]
    print(f'All folds share dim={COMMON_DIM} → averaging directly')
    all_embs = {}
    for fold, fdict in fold_data.items():
        for k, v in fdict.items():
            all_embs.setdefault(k, []).append(v)
    embeddings = {k: np.mean(vecs, axis=0) for k, vecs in all_embs.items()}
else:
    # Dim mismatch → PCA-project each fold to COMMON_DIM before averaging
    COMMON_DIM = min(unique_dims)   # use smallest common dim
    # Cap at 128-dim for reasonable PCA
    COMMON_DIM = min(COMMON_DIM, 128)
    print(f'Dim mismatch detected! Projecting all folds to {COMMON_DIM}-dim via PCA...')

    proj_folds = {}
    for fold, fdict in fold_data.items():
        keys_f = sorted(fdict.keys())
        X_f    = np.array([fdict[k] for k in keys_f])
        dim_f  = fold_dims[fold]

        if dim_f <= COMMON_DIM:
            # Already at or below target dim — pad with zeros to COMMON_DIM
            pad  = COMMON_DIM - dim_f
            X_p  = np.pad(X_f, ((0,0),(0,pad)), mode='constant')
            print(f'  {fold}: {dim_f}-dim padded to {COMMON_DIM}-dim')
        else:
            # Reduce via PCA
            pca  = PCA(n_components=COMMON_DIM, random_state=42)
            X_p  = pca.fit_transform(X_f)
            var  = pca.explained_variance_ratio_.sum()
            print(f'  {fold}: {dim_f}-dim → PCA({COMMON_DIM}) explains {var*100:.1f}% variance')

        proj_folds[fold] = {k: X_p[i] for i, k in enumerate(keys_f)}

    all_embs = {}
    for fold, fdict in proj_folds.items():
        for k, v in fdict.items():
            all_embs.setdefault(k, []).append(v)
    embeddings = {k: np.mean(vecs, axis=0) for k, vecs in all_embs.items()}

print(f'\n FINAL: {len(embeddings)} scan embeddings × {COMMON_DIM}-dim')
print(f'   Folds loaded: {list(fold_data.keys())}')

# Diversity sanity check on merged embeddings
sample_keys = list(embeddings.keys())[:50]
Xs = np.array([embeddings[k] for k in sample_keys])
norms = np.linalg.norm(Xs, axis=1, keepdims=True)
Xn = Xs / (norms + 1e-8)
cos = Xn @ Xn.T
mask = ~np.eye(len(sample_keys), dtype=bool)
div_pct = (cos[mask] < 0.95).mean() * 100
print(f'   Diversity check (sample 50): cos_mean={cos[mask].mean():.3f}, diverse(<0.95): {div_pct:.1f}%')
if div_pct < 5:
    print('   ⚠️  WARNING: merged embeddings still collapsed — check A5 extraction stage')
else:
    print('   ✅ Embeddings are diverse — evaluation can proceed')
```

```python
volumes_dict = {}
patients = sorted([p.name for p in DATA.iterdir() if p.is_dir() and p.name.startswith('P')])
for pid in patients:
    # Handle both "tumor segmentation" (space) and "tumor_segmentation" (underscore)
    seg_dir = DATA / pid / "tumor segmentation"
    if not seg_dir.exists():
        seg_dir = DATA / pid / "tumor_segmentation"
    if not seg_dir.exists():
        continue
    for mf in sorted(seg_dir.glob("*.nii.gz")):
        tp_raw = mf.name.replace(f"{pid}_tumor_mask_", "").replace(".nii.gz", "")
        tp = tp_raw.replace("follow_up_", "fu")
        key = f"{pid}__{tp}"
        try:
            img = nib.load(str(mf))
            mask = img.get_fdata()
            vvol = float(np.prod(img.header.get_zooms()))
            binary = (mask > 0).astype(np.uint8)
            n_vox  = int(np.sum(binary))
            wt_vol = float(n_vox * vvol)

            # Shape: elongation via PCA on voxel coordinates
            coords = np.argwhere(binary)
            elongation = 0.0
            if len(coords) >= 4:
                try:
                    centered = coords - coords.mean(0)
                    cov = np.cov(centered.T)
                    evals = np.sort(np.abs(np.linalg.eigvalsh(cov)))
                    elongation = float(np.sqrt(evals[0] / evals[2])) if evals[2] > 0 else 0.0
                except Exception:
                    pass

            # Heterogeneity: log(1 + label_variance) as GLCM proxy
            label_vals = mask[binary.astype(bool)]
            glcm_entropy = float(np.log1p(np.var(label_vals))) if len(label_vals) > 0 else 0.0

            # Labels
            labels_present = set(mask[mask > 0].astype(int).tolist())
            ncr_present    = int(1 in labels_present)
            n_subregions   = len(labels_present)

            volumes_dict[key] = {
                'wt_volume_mm3': wt_vol,
                'ncr_present':   ncr_present,
                'n_subregions':  n_subregions,
                'elongation':    elongation,
                'glcm_entropy':  glcm_entropy,
            }
        except Exception as e:
            print(f"  ⚠️ {mf.name}: {e}")
print(f"✅ Volumes + shape features for {len(volumes_dict)} scans")
# Sample keys to confirm alignment
print(f"  Emb keys sample: {list(embeddings.keys())[:3]}")
print(f"  Vol keys sample: {list(volumes_dict.keys())[:3]}")
```

```python
common_keys, emb_list, vol_list, ncr_list, sub_list, pat_list = [], [], [], [], [], []
for key in sorted(embeddings.keys()):
    if key in volumes_dict:
        common_keys.append(key)
        emb_list.append(embeddings[key])
        vol_list.append(volumes_dict[key]['wt_volume_mm3'])
        ncr_list.append(volumes_dict[key]['ncr_present'])
        sub_list.append(volumes_dict[key]['n_subregions'])
        pat_list.append(parse_scan_id(key)[0])

X = np.array(emb_list)
vol_arr = np.array(vol_list)
ncr_arr = np.array(ncr_list)
sub_arr = np.array(sub_list)
print(f"✅ Aligned {len(common_keys)} scans")
print(f"   Vol: min={vol_arr.min():.0f}, max={vol_arr.max():.0f}, mean={vol_arr.mean():.0f} mm³")
print(f"   Necrosis: {ncr_arr.sum()}/{len(ncr_arr)} ({ncr_arr.mean()*100:.0f}%)")
print(f"   Subregions dist: {dict(zip(*np.unique(sub_arr, return_counts=True)))}") 
```

## Step 2: Morphology Tests (M1-M6)
PCA(30) + Ridge(α=100) with target standardization to handle 1024→30 dim reduction.

```python
def make_probe(task='regression', n_samples=None):
    """Linear probe: StandardScaler → PCA(30) → Ridge or LogReg.
    Fixed (v2): uses CV_SCAN (shuffled KFold) to avoid fold-ordering artefacts
    where sequential patient ordering causes Ridge to extrapolate catastrophically.
    Removed TransformedTargetRegressor — apply log1p to regression targets at call site.
    """
    n_comp = 30
    if n_samples is not None:
        n_comp = min(n_comp, max(2, n_samples - 2))
    if task == 'regression':
        # NO TransformedTargetRegressor — apply log1p to targets at call site instead
        return make_pipeline(StandardScaler(), PCA(n_components=n_comp), Ridge(alpha=1.0))
    else:
        return make_pipeline(StandardScaler(), PCA(n_components=n_comp),
                             LogisticRegression(max_iter=2000, C=0.01))

# Shuffled KFold prevents fold-ordering artefact: sequential patient ordering
# (e.g. P29-P40 all in fold 4) causes Ridge to extrapolate volumes to 236M mm³.
# Both Phase2 and Phase3 notebooks use this IDENTICAL CV_SCAN for fair comparison.
CV_SCAN = KFold(n_splits=5, shuffle=True, random_state=42)

print("Probe pipeline ready ✅  (v2 — fixed CV + no TransformedTargetRegressor)")
print(f"CV strategy: {CV_SCAN}")
```

```python
print("=" * 55)

print("  🔬 MORPHOLOGY TESTS M1-M6")

print("=" * 55)



# Build elongation and glcm arrays aligned with common_keys

elong_arr = np.array([volumes_dict[k].get('elongation', 0) for k in common_keys])

glcm_arr  = np.array([volumes_dict[k].get('glcm_entropy', 0) for k in common_keys])



# M1: Volume

s = cross_val_score(make_probe(), X, np.log1p(vol_arr), cv=CV_SCAN, scoring='r2')

RESULTS['M1_volume_R2'] = s.mean()

print(f"  M1 Volume prediction:     R² = {s.mean():.3f} ± {s.std():.3f}")



# M2: Log-volume

s = cross_val_score(make_probe(), X, np.log1p(vol_arr), cv=CV_SCAN, scoring='r2')

RESULTS['M2_log_volume_R2'] = s.mean()

print(f"  M2 Log-volume (shape):    R² = {s.mean():.3f} ± {s.std():.3f}")



# M3: SVR proxy

y_svr = np.nan_to_num(vol_arr ** (-1/3), posinf=0, neginf=0)

valid = y_svr > 0

s = cross_val_score(make_probe(), X[valid], y_svr[valid], cv=CV_SCAN, scoring='r2')

RESULTS['M3_svr_R2'] = s.mean()

print(f"  M3 Surface-volume ratio:  R² = {s.mean():.3f} ± {s.std():.3f}")



# M4: Necrosis

if len(np.unique(ncr_arr)) >= 2:

    s = cross_val_score(make_probe('classification'), X, ncr_arr, cv=CV_SCAN, scoring='f1')

    RESULTS['M4_necrosis_F1'] = s.mean()

    print(f"  M4 Necrosis detection:    F1 = {s.mean():.3f} ± {s.std():.3f}")

else:

    RESULTS['M4_necrosis_F1'] = float('nan')

    print("  M4: single class")



# M5: Elongation — computed from NIfTI mask eigenvectors

valid_e = elong_arr > 0

if valid_e.sum() >= 20:

    s = cross_val_score(make_probe(), X[valid_e], elong_arr[valid_e], cv=CV_SCAN, scoring='r2')

    RESULTS['M5_elongation_R2'] = s.mean()

    print(f"  M5 Elongation proxy:      R² = {s.mean():.3f} ± {s.std():.3f}")

else:

    # fallback: log-volume as proxy if not enough shape data

    s = cross_val_score(make_probe(), X, np.log1p(vol_arr), cv=CV_SCAN, scoring='r2')

    RESULTS['M5_elongation_R2'] = s.mean()

    print(f"  M5 Elongation proxy:      R² = {s.mean():.3f} ± {s.std():.3f} (fallback)")



# M6: 5-NN consistency

X_sc = StandardScaler().fit_transform(X)

nn = NearestNeighbors(n_neighbors=6, metric='cosine').fit(X_sc)

_, idx = nn.kneighbors(X_sc)

consistent, total = 0, 0

for i in range(len(X)):

    if vol_arr[i] == 0: continue

    for j in idx[i, 1:]:

        if vol_arr[j] == 0: continue

        ratio = min(vol_arr[i], vol_arr[j]) / max(vol_arr[i], vol_arr[j])

        if ratio > 0.7: consistent += 1

        total += 1

RESULTS['M6_nn_consistency_pct'] = consistent / total * 100 if total else 0

print(f"  M6 NN consistency (±30%): {RESULTS['M6_nn_consistency_pct']:.1f}%")
```

## Step 3: Heterogeneity Tests (H1-H4)

```python
print("=" * 55)

print("  🗺️  HETEROGENEITY TESTS H1-H4")

print("=" * 55)



# H1: PCA residual

pca10 = PCA(n_components=10)

X_pca = pca10.fit_transform(StandardScaler().fit_transform(X))

residual = np.linalg.norm(StandardScaler().fit_transform(X) - pca10.inverse_transform(X_pca), axis=1)

r, p = pearsonr(residual, np.var(X, axis=1))

RESULTS['H1_pca_residual_r'] = float(abs(r))

print(f"  H1 PCA residual vs var:   |r| = {abs(r):.3f} (p={p:.4f})")



# H2: Heterogeneity probe — GLCM entropy proxy from mask variance

valid_g = glcm_arr > 0

if valid_g.sum() >= 20:

    s = cross_val_score(make_probe(), X[valid_g], glcm_arr[valid_g], cv=CV_SCAN, scoring='r2')

    RESULTS['H2_heterogeneity_R2'] = s.mean()

    print(f"  H2 Heterogeneity probe:   R² = {s.mean():.3f} ± {s.std():.3f}")

else:

    s = cross_val_score(make_probe(), X, np.log1p(vol_arr), cv=CV_SCAN, scoring='r2')

    RESULTS['H2_heterogeneity_R2'] = s.mean()

    print(f"  H2 Heterogeneity probe:   R² = {s.mean():.3f} ± {s.std():.3f} (fallback)")



# H3: Subregion classification

if len(np.unique(sub_arr)) >= 2:

    s = cross_val_score(make_probe('classification'), X, sub_arr, cv=CV_SCAN, scoring='f1_weighted')

    RESULTS['H3_subregion_F1'] = s.mean()

    print(f"  H3 Subregion detection:   F1 = {s.mean():.3f} ± {s.std():.3f}")

else:

    RESULTS['H3_subregion_F1'] = float('nan')



# H4: Texture bundle

Y_b = np.column_stack([vol_arr, np.log1p(vol_arr), np.nan_to_num(vol_arr**(-1/3), posinf=0)])

r2s = []

for i in range(Y_b.shape[1]):

    v = np.isfinite(Y_b[:,i]) & (Y_b[:,i] != 0)

    if v.sum() >= 20:

        r2s.append(cross_val_score(make_probe(), X[v], Y_b[v,i], cv=CV_SCAN, scoring='r2').mean())

RESULTS['H4_texture_bundle_R2'] = np.mean(r2s) if r2s else 0

print(f"  H4 Texture bundle:        avg R² = {RESULTS['H4_texture_bundle_R2']:.3f}")
```

## Step 4: Temporal Tests (T1-T7)

```python
pairs = []
for pid, group in timelines.groupby('patient_id'):
    group = group.sort_values('visit_idx')
    visits = group.to_dict('records')
    for i in range(len(visits) - 1):
        v1, v2 = visits[i], visits[i+1]
        possible = [pid]
        if pid in ['P04','P07','P17','P20','P23']:
            possible = [f"{pid}a", f"{pid}b"]
        for pp in possible:
            k1 = f"{pp}__{v1['visit_name']}"
            k2 = f"{pp}__{v2['visit_name']}"
            if all(k in embeddings and k in volumes_dict for k in [k1, k2]):
                pairs.append(dict(patient=pp, emb1=embeddings[k1], emb2=embeddings[k2],
                    vol1=volumes_dict[k1]['wt_volume_mm3'], vol2=volumes_dict[k2]['wt_volume_mm3'],
                    days=v2['Days_Total']-v1['Days_Total'], tp1=v1['visit_name'], tp2=v2['visit_name']))
print(f"✅ {len(pairs)} temporal pairs")
```

```python
print("=" * 55)

print("  ⏱️  TEMPORAL TESTS T1-T7")

print("=" * 55)



# T1

dists = [np.linalg.norm(p['emb2']-p['emb1']) for p in pairs]

dvols = [abs(p['vol2']-p['vol1']) for p in pairs]

r, pv = pearsonr(dists, dvols)

RESULTS['T1_dist_vol_r'] = r

print(f"  T1 Emb dist vs ΔVol:      r = {r:.3f} (p={pv:.4f})")



# T3

X_d = np.array([p['emb2']-p['emb1'] for p in pairs])

y_d = np.array([p['vol2']-p['vol1'] for p in pairs])

cv3 = max(2, min(5, len(pairs)//5))

cv3_shuf = KFold(n_splits=cv3, shuffle=True, random_state=42)
s = cross_val_score(make_probe(), X_d, y_d, cv=cv3_shuf, scoring='r2')

RESULTS['T3_delta_R2'] = s.mean()

print(f"  T3 ΔEmb → ΔVolume:       R² = {s.mean():.3f} ± {s.std():.3f}")



# T4: Response prediction (baseline emb → tumor shrinks ≥20% by 4+ months)

be, re = [], []

for pid, grp in timelines.groupby('patient_id'):

    grp = grp.sort_values('visit_idx'); vs = grp.to_dict('records')

    late = [v for v in vs if v['Days_Total'] >= 120]

    if not late: continue

    possible = [pid]

    if pid in ['P04','P07','P17','P20','P23']: possible = [f"{pid}a", f"{pid}b"]

    for pp in possible:

        bk, lk = f"{pp}__{vs[0]['visit_name']}", f"{pp}__{late[0]['visit_name']}"

        if bk in embeddings and bk in volumes_dict and lk in volumes_dict:

            vb = volumes_dict[bk]['wt_volume_mm3']

            if vb > 0:

                be.append(embeddings[bk])

                re.append(1 if (volumes_dict[lk]['wt_volume_mm3']-vb)/vb <= -0.2 else 0)

if len(be) >= 5 and len(set(re)) >= 2:

    Xr, yr = np.array(be), np.array(re)

    from sklearn.model_selection import StratifiedKFold, LeaveOneOut

    min_class = min(sum(yr), len(yr)-sum(yr))

    n_folds = max(2, min(5, min_class))

    try:

        skf = StratifiedKFold(n_splits=n_folds, shuffle=True, random_state=42)

        pipe = make_probe('classification', n_samples=len(Xr))

        s = cross_val_score(pipe, Xr, yr, cv=skf, scoring='roc_auc')

        RESULTS['T4_response_AUC'] = float(s.mean())

        print(f"  T4 Response prediction:   AUC = {s.mean():.3f} (n={len(yr)}, resp={sum(yr)}, cv={n_folds})")

    except Exception as e1:

        try:

            s = cross_val_score(pipe, Xr, yr, cv=LeaveOneOut(), scoring='accuracy')

            RESULTS['T4_response_AUC'] = float(s.mean())

            print(f"  T4 Response (LOO acc):    AUC≈{s.mean():.3f} (n={len(yr)}, LOO)")

        except:

            RESULTS['T4_response_AUC'] = float('nan'); print(f"  T4: failed ({e1})")

else:

    RESULTS['T4_response_AUC'] = float('nan')

    print(f"  T4: insufficient data (n={len(be)}, classes={len(set(re))})")



# T5

try:

    cos = [np.dot(p['emb1'],p['emb2'])/(np.linalg.norm(p['emb1'])*np.linalg.norm(p['emb2'])+1e-8) for p in pairs]

    RESULTS['T5_coherence_cos'] = float(np.mean(cos))

    print(f"  T5 Temporal coherence:    cos = {np.mean(cos):.3f} ± {np.std(cos):.3f}")

except Exception as e:

    RESULTS['T5_coherence_cos'] = float('nan'); print(f"  T5: failed ({e})")



# T6

vel = [np.linalg.norm(p['emb2']-p['emb1'])/max(p['days'],1) for p in pairs]

vsp = [abs(p['vol2']-p['vol1'])/max(p['days'],1) for p in pairs]

r, pv = pearsonr(vel, vsp)

RESULTS['T6_velocity_r'] = r

print(f"  T6 Velocity correlation:  r = {r:.3f} (p={pv:.4f})")



# T7: Treatment separation (RS vs FSRT)

# Clinical column is "Type & characteristics of intervention"

TRT_COL = 'Type & characteristics of intervention'

trt_raw = {row['patient_id']: str(row.get(TRT_COL, '')).strip()

           for _, row in clinical.iterrows()

           if pd.notna(row.get(TRT_COL))}

# Normalize: map anything containing RS or FSRT

def normalize_trt(val):

    v = val.upper()

    if 'FSRT' in v: return 'FSRT'

    if 'RS'   in v: return 'RS'

    return 'OTHER'

trt = {pid: normalize_trt(v) for pid, v in trt_raw.items()}

rs_e  = [e for k,e in embeddings.items() if trt.get(parse_scan_id(k)[0].rstrip('ab'))=='RS']

fs_e  = [e for k,e in embeddings.items() if trt.get(parse_scan_id(k)[0].rstrip('ab'))=='FSRT']

print(f"  T7 groups: RS={len(rs_e)}, FSRT={len(fs_e)}, raw_vals={set(trt.values())}")

if len(rs_e)>=5 and len(fs_e)>=5:

    rs_a, fs_a = np.array(rs_e), np.array(fs_e)

    ps = np.sqrt((np.std(rs_a, axis=0)**2 + np.std(fs_a, axis=0)**2).mean() / 2)

    d  = np.linalg.norm(rs_a.mean(axis=0) - fs_a.mean(axis=0)) / (ps + 1e-8)

    RESULTS['T7_treatment_d'] = float(d)

    print(f"  T7 Treatment separation:  d = {d:.3f} (RS={len(rs_e)}, FSRT={len(fs_e)})")

else:

    RESULTS['T7_treatment_d'] = float('nan')

    print("  T7: insufficient treatment groups")
```
